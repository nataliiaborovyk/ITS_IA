# **Laboratorio di ML**:  applicazione degli argomenti visti in teoria

In questa lezione svolgeremo una attività laboratoriale di applicazione dei metodi visti fino a questo momento.


1) Scegli un dataset e studia il problema da qui: https://archive.ics.uci.edu/

2) fai una analisi esplorativa dei dati (EDA)

In particolare fai attenzione ai seguenti aspetti: tipologia di variabili presenti (continue, categoriche, eccc...) problema dei missing data (NAN, NA, ecc..) e il problema del preprocessing delle variabili(feature preprocessing)

3) Fare in preprocessing e imputazione dei dati mancanti (Missing Value)


3) applica le procedure viste di

- separazione training e test (prova con e senza cross validation)
- creazione del modello di classificatore con i relativi paramentri
- applicazione del modello (fit)
- estrapolazione delle principali metriche (accuracy, confusion matrix, curva roc, curva precision e recall, area sotto le curve)
- riflessione su valori ottenuti e come migliorarli
